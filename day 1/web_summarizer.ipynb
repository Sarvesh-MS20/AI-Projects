```python
# imports

import os
from groq import Groq
from dotenv import load_dotenv
from webscraping.scraper import fetch_website_contents
from IPython.display import display,Markdown
```

```python
# getting the api key
load_dotenv(override=True)
api_key = os.getenv("api_key")

try:
    if not api_key:
        raise ValueError("Api key not found")
    else:
        print("api key found")

except ValueError as e:
    print(e)
```

```python
# setting up the model connecting to Groq

client = Groq(api_key=api_key)
```

```python
system_prompt = """
You are an AI asssistant that analyses
the important information from a website and gives a short summary of the text,
ignoring the text that might be navigation related
Respond in markdown
Do not wrap the markdown in the code block - respond just with the markdown
"""

user_prompt = """
Here are the content of a website 
Provide a short summary of the website.
if it has news or announcements summarize that too.
Respond in markdown
"""
```

```python

def messages_for(website):
    messages = [{"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt + website}
                ]
    return messages

def summarizer(url):
    scrape = fetch_website_contents(url)
    response = client.chat.completions.create(
        model= "openai/gpt-oss-120b",
        messages=messages_for(scrape),
        temperature=0.5
    )

    output = response.choices[0].message.content
    return output

#print(summarizer("https://edwarddonner.com"))

# A fucntion to display everything in an elegant way using markdown

def MarkdownDisplay(url):
    summary = summarizer(url)
    display(Markdown(summary))

MarkdownDisplay("https://edwarddonner.com")
```
